### easy problem sum of two digits

### Given a number n , find the sum of its digits 
**Examples : **

input : n = 687 
output: 21

## Digit Extration O(log10n)time and O(1) Space. 

```python
def sumOfDigits(n)-> int:
    sum =0 
    while n!=0: 
         sum+= n%10
         #update n 
         n=int(n/10)

    return sum
```

```python
if __name__=="__main__":
    n = 687
    print(sumOfDigits(n))
```

```python
 def SOD(n) -> int :
     if n==0 :
         return 0 
     return n%10 + SOD(n//10)
     
```
